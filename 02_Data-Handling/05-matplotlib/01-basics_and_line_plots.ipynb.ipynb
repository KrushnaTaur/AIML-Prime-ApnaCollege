```python
#Real Data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M
```

```python
plt.plot(years, oscar_revenue)
plt.show()
```

```python
plt.plot(years, oscar_revenue)
plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.show()
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M
```

```python
plt.plot(years, oscar_revenue)
plt.plot(years, non_oscar_revenue)

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.show()
```

```python
plt.plot(years, oscar_revenue, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

```python
plt.plot(years, oscar_revenue, "o--r", label="Oscar Movies")
plt.plot(years, non_oscar_revenue,"<-y", label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

```python
plt.plot(years, oscar_revenue, color="red", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, color="green", label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

```python
plt.plot(years, oscar_revenue, color="red",marker="<", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, color="green",marker="o", label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

```python
plt.plot(years, oscar_revenue, color="red", marker="<", linestyle="--", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, color="green", marker="o", linestyle="-", label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

```python
plt.plot(years, oscar_revenue, color="red", marker="<", linestyle="--", linewidth="5", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, color="green", marker="o", linestyle="-", linewidth="3", label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```
