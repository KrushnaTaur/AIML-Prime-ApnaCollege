```python
import matplotlib.pyplot as plt
```

```python
# Bar charts

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M
```

```python
plt.bar(years, oscar_revenue)

plt.title("Revenue in each year for Oscar Movies")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.show()
```

```python
plt.bar(years, oscar_revenue, color="yellow")

plt.title("Revenue in each year for Oscar Movies")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.show()
```

```python
plt.bar(years, oscar_revenue, color="green")

plt.title("Revenue in each year for Oscar Movies")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.text(years[0], oscar_revenue[0], str(oscar_revenue[0]), ha="center")

plt.show()
```

```python
plt.bar(years, oscar_revenue, color="green")

plt.title("Revenue in each year for Oscar Movies")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i]+10, str(oscar_revenue[i]), ha="center")

plt.show()
```

```python
plt.bar(years, oscar_revenue, color="green")

plt.title("Revenue in each year for Oscar Movies")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i]+10, str(oscar_revenue[i]), ha="center")

plt.ylim(0, max(oscar_revenue)+100)

plt.show()
```

```python
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]
```

```python
plt.bar(years, oscar_revenue)
plt.bar(years, non_oscar_revenue)

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.show()
```

```python
import numpy as np
np.arange(len(years))
```

```python
x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="Oscar Movies")
plt.bar(x+width/2, non_oscar_revenue, width, label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

```python
x = np.arange(len(years))
width = 0.25

plt.bar(x-width, oscar_revenue, width, label="Oscar Movies")
plt.bar(x, non_oscar_revenue, width, label="Non-Oscar Movies")
plt.bar(x+width, oscar_revenue, width, label="Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

```python
x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="Oscar Movies")
plt.bar(x+width/2, non_oscar_revenue, width, label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.xticks(x, years)

plt.show()
```

```python
# Pie Charts

expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
```

```python
plt.pie(amounts, labels=expenses)

plt.show()
```

```python
plt.style.use("dark_background")

plt.pie(amounts, labels=expenses)

plt.show()
```

```python
plt.style.use("default")

plt.pie(amounts, labels=expenses, autopct="%1.1f%%")

plt.show()
```

```python
colors = ["red", "black", "pink", "blue", "yellow"]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%", colors=colors)

plt.show()
```

```python
plt.style.use("default")

plt.pie(amounts, labels=expenses, autopct="%1.1f%%")

plt.show()
```

```python
plt.style.use("default")

plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
       wedgeprops= {
           "edgecolor": "black",
           "linewidth": 1,
           "linestyle": "--"
       }
       )

plt.show()
```

```python
explode = [0, 0, 0, 0.2, 0]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
       wedgeprops= {
           "edgecolor": "black",
           "linewidth": 1,
           "linestyle": "--"
       },
        explode=explode
       )

plt.show()
```

```python
explode = [0, 0, 0, 0.2, 0]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
       wedgeprops= {
           "edgecolor": "black",
           "linewidth": 1,
           "linestyle": "--"
       },
        explode=explode,
        startangle=90
       )

plt.show()
```

```python
explode = [0, 0, 0, 0.2, 0]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
       wedgeprops= {
           "edgecolor": "black",
           "linewidth": 1,
           "linestyle": "--"
       },
        explode=explode,
        startangle=180
       )

plt.show()
```

```python
explode = [0, 0, 0, 0.2, 0]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
       wedgeprops= {
           "edgecolor": "black",
           "linewidth": 1,
           "linestyle": "--"
       },
        explode=explode,
        startangle=360,
        shadow=True
       )

plt.show()
```
