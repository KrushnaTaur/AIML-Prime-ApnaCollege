```python
import numpy as np
```

```python
#Vectorization

arr = np.array([1, 2, 3, 4, 5])
print(arr ** 2)
print(arr + 10)
```

```python
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([5, 4, 3, 2, 1])

print(arr1 + arr2)
print(arr1 - arr2)
```

```python
#Brodcasting - (scaling)

arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([[5, 4, 3, 2, 1], [1, 2, 3, 4, 5]])

print(arr1.shape)
print(arr2.shape)
print(arr1 + arr2)
```

```python
#normalize

arr = np.array([[1, 2], [3, 4]])

mean = np.mean(arr)
std = np.std(arr)

print(mean)
print(std)
```

```python
normalized_arr = (arr - mean) / std

print(normalized_arr)
```
