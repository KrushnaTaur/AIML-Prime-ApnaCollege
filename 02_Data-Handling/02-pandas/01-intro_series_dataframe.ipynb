```python
import pandas as pd
```

```python
#Series

s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))
```

```python
print(s[0])
print(s[2])
```

```python
s = pd.Series([23, 24, 25, 26], index=["Adam", "Bob", "Charlie", "Dude"])
print(s)
```

```python
print(s["Adam"])
print(s["Dude"])
```

```python
print(s.index)
```

```python
#Dataframe

info = {
    "Name" : ["Krushna", "Ketan", "Rohit"],
    "Age" : [20, 21, 19],
    "CGPA" : [8.32, 8.0, 7.75]
}

df = pd.DataFrame(info)
print(df)
print(type(df))
print(df.index)
print(df.columns)
```

```python
df
```

```python
df = pd.DataFrame([["Adam", 23], ["Bob", 24], ["Charlie", 25]], columns=["Name", "Age"])

print(df)
```

```python
df
```

```python
import numpy as np

np_arr = np.array([[1, 2, 3], [4, 5, 6]])

df = pd.DataFrame(np_arr, columns=["A", "B", "C"])

print(df)
```
