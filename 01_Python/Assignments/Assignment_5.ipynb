Q1. Create a program that:
1. Opens a file"names.txt"in write mode
2. Writes 5 names (one per line) entered by the user
3. Then opens the same file in read mode and prints all names

```python
with open("names.txt", "w") as file:
    print("Enter 5 names:")
    for i in range(5):
        name = input(f"Enter name {i+1}: ")
        file.write(name + "\n")
```

```python
print("Names stored in the file:")
with open("names.txt", "r") as file:
    for line in file:
        print(line.strip())
```

Q2. Create a program that:
1. Opens a file  "log.txt"in append mode
2. Adds a new log entry (like "Program run successfully")
3. Opens the file in read mode and prints all logs

```python
with open("log.txt", "a") as file:
    file.write("Program started\n")
    file.write("User logged in\n")
    file.write("Program run successfully\n")
```

```python
print("Log File Contents:")
with open("log.txt", "r") as file:
    for line in file:
        print(line.strip())
```

Q3.Create a program that:
1. Has a list of numbers: [5, 10, 15, 20, 25]
2. Uses a list comprehension to create a new list with only numbers greater
than 15
3. Prints the new list

```python
numbers = [5, 10, 15, 20, 25]

filtered_numbers = [num for num in numbers if num > 15]

print("Numbers greater than 15:", filtered_numbers)
```

Q4. Create a Python dictionary of 3 cities and their populations. Save it to "cities.json".
1. Then load the JSON and print each city and its population.
2. Ask the user for a new city & its population  - update this info in the json file.

```python
import json

# Step 1: Create dictionary
cities = {
    "Mumbai": 20411000,
    "Delhi": 16787941,
    "Bangalore": 12425300
}
```

```python
with open("cities.json", "w") as file:
    json.dump(cities, file, indent=4)
```

```python
with open("cities.json", "r") as file:
    data = json.load(file)

print("Cities and their populations:")
for city, population in data.items():
    print(f"{city}: {population}")
```

```python
new_city = input("Enter new city name: ")
new_population = int(input("Enter population: "))

data[new_city] = new_population
```

```python
with open("cities.json", "w") as file:
    json.dump(data, file, indent=4)

print("Updated data saved to cities.json")
```

Q5. Write a program that tries to open "data.txt"in read mode. If the file does not exist, catch the exception and print "File not found!"

```python
try:
    with open("data.txt", "r") as file:
        content = file.read()
        print("File contents:")
        print(content)

except FileNotFoundError:
    print("File not found!")
```
